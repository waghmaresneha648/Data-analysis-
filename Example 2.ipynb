```python
from collections import Counter
import re


def calculate_word_frequency(file_path: str) -> dict:
 
  
  try:
    # Open and read the file safely
    with open(file_path, 'r', encoding='utf-8') as file:
      text = file.read()

    # Convert to lowercase and extract words using a regular expression
    words = re.findall(r'\b\w+\b', text.lower())

    # Count the frequencies using Counter and convert back to a standard dict
    return dict(Counter(words))

  except FileNotFoundError:
    print(f'Error: The file "{file_path}" was not found.')
    return {}


```
